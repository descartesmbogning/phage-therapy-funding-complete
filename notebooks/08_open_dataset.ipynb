# 08 · Open dataset and explorer data

Writes the public, aggregate files that go with the paper:

- `outputs/explorer/cube.csv` and `meta.json`: the data behind the interactive explorer (cognosx.ca/phage-funding), by year, funder country, funder, recipient country, income group and topic, with fractional counting. The build stops if its totals differ from the paper.
- `docs/data_dictionary.csv`: one row per column of every public CSV (file, column, type, unit, description).
- `datapackage.json`: a Frictionless Data Package describing the same files, so tools can load and validate them.

Nothing here contains researcher names, grant titles or grant IDs. See `docs/DATASHEET.md` for what the data can and cannot be used for.

```python
import sys
from pathlib import Path
import pandas as pd
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT / "src"))
import open_data as od

d = pd.read_csv(ROOT / "data" / "processed" / "phage_funding_analysis.csv")
summary = od.write_all(ROOT, d, version="1.0.0")
summary
```

```python
cube = pd.read_csv(ROOT / "outputs" / "explorer" / "cube.csv")
print(f"{len(cube)} rows; {cube.funder.nunique()} funders; {cube.recipient_country.nunique()} recipient countries")
cube.groupby("recipient_income").usd.sum().div(cube.usd.sum()).mul(100).round(2).rename("% of recorded funding")
```
